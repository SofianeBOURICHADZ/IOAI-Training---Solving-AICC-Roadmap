```python
import pandas as pd
import numpy as np

train = pd.read_csv('train_data.csv')
test = pd.read_csv('test_data.csv')
```

```python
subtask1_answer = len(test[(test['Sex'] == 'Femelă') & (test['Ureche_lăsată'] == True) & (test['Culoare'] == 'Havana')])
```

```python
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import OneHotEncoder
from sklearn.decomposition import TruncatedSVD,PCA
from sklearn.cluster import KMeans
pip = Pipeline([
    ('ohe', OneHotEncoder(sparse_output=True, handle_unknown='ignore')),
    ('svd', PCA(n_components=25, random_state=42)),
    ('kmeans', KMeans(n_clusters=3, n_init=10, random_state=42))
])

subtask2_answers = pip.fit_predict(test.drop(['ID'],axis=1))
```

```python
from catboost import CatBoostRegressor

catboost = CatBoostRegressor(iterations = 300, rsm = 0.85, max_depth = 6, l2_leaf_reg = 0.2, random_state = 42,cat_features = ['Sex','Ureche_lăsată','Culoare','Tip_blană','Calitate_blană','Formă_corp','Apare_gușa','Sănătate'])
catboost.fit(train.drop(['ID','Scor_jurizare'],axis=1), train['Scor_jurizare'])

subtask3_answers = catboost.predict(test.drop('ID',axis=1))
```

```python
t1 = pd.DataFrame({
    'subtaskID':[1],'datapointID':[1],'answer':[subtask1_answer]
})
t2 = pd.DataFrame({
    'subtaskID':2,'datapointID':test['ID'].values,'answer':subtask2_answers
})
t3 = pd.DataFrame({
    'subtaskID':3,'datapointID':test['ID'].values,'answer':subtask3_answers
})

pd.concat([t1,t2,t3]).to_csv('submission.csv',index=False)
```
