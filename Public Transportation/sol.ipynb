```python
import numpy as np
import pandas as pd

data = pd.read_csv('data.csv')
```

```python
subtask1_answer = [len(data['id'].unique()),len(data['vehicle_type'].unique())]
```

```python
from sklearn.cluster import KMeans

##USE PLT TO MAP THE COORDINATES With matplotlib.pyploy.scatter and you will see 3 masses

kmeans = KMeans(n_clusters = 3, random_state = 42)
subtask2_answer = kmeans.fit_predict(data[['latitude','longitude']].to_numpy())
```

```python
s3 = data[(data['vehicle_type'] == 10)]
ids = []
for i in range(0,len(s3)):
    if ((int(s3['timestamp'].iloc[i].split()[0].split('-')[2]) == 12) and (int(s3['timestamp'].iloc[i].split()[1].split(':')[0]) >= 21) and (int(s3['timestamp'].iloc[i].split()[1].split(':')[1]) >= 40)):
        ids.append(i)
```

```python
depots = KMeans(n_clusters = 3).fit(s3[['latitude', 'longitude']].to_numpy()).cluster_centers_
depots = depots[np.argsort(depots[:,0])]
```

```python
pd.DataFrame({
    'subtaskID':[1] + [2] * len(data) + [3] * 3,
    'Value1' : [subtask1_answer[0]] +[int(x) for x in data['id'].astype('int').tolist()] + list(depots[:,0]),
    'Value2':[subtask1_answer[1]] + [int(X) for X in subtask2_answer] + list(depots[:,1])
}).to_csv('submission.csv',index=False)
```
