```python
import numpy as np
import pandas as pd

train = pd.read_csv('train_data.csv')
test = pd.read_csv('test_data.csv')
```

```python
subtask1_answer = [(test['Square_Footage'].iloc[x] + test['Garage_Size'].iloc[x] + test['Lot_Size'].iloc[x]) for x in range(0, len(test))]
```

```python
subtask2_answer = [(test['Garage_Size'].iloc[x] / test['Total_Rooms'].iloc[x] ) for x in range(0, len(test))]
```

```python
subtask3_answer = [((test['Solar_Exposure_Index'].iloc[x] - test['Vibration_Level'].iloc[x]) / test['Magnetic_Field_Strength'].iloc[x]) for x in range(0, len(test))]
```

```python
house_average = train['Square_Footage'].mean()
subtask4_answer = (test['Square_Footage'] - house_average).tolist()
```

```python
from catboost import CatBoostRegressor
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
price_mean = train['Price'].mean()
price_std= train['Price'].std()

model = CatBoostRegressor(iterations = 700, learning_rate = 0.17, max_depth = 6, l2_leaf_reg = 0.2, rsm = 0.85, random_state = 64,cat_features = ['Num_Bedrooms','Num_Bathrooms','Neighborhood_Quality'])
model.fit(scaler.fit_transform(train.drop(['ID','Price','Year_Built','Total_Rooms'],axis=1).to_numpy()),((train['Price'] - price_mean) / price_std).to_numpy())
```

```python
subtask5_answers = model.predict(scaler.transform(test.drop('ID',axis=1).to_numpy()))
```

```python
sds = []
dps = []
ans = []
for i in range(0,len(test)):
    sds.extend([1,2,3,4,5])
    dps.extend([test['ID'].iloc[i]] * 5)
    ans.extend([subtask1_answer[i], subtask2_answer[i], subtask3_answer[i], subtask4_answer[i], subtask5_answers[i]])

pd.DataFrame({
    'subtaskID':sds,
    'datapointID':dps,
    'answer':ans
}).to_csv('submission.csv',index=False)
```
