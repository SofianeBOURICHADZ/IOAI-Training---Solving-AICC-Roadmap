```python
import pandas as pd
import numpy as np

train = pd.read_csv('train.csv')
test = pd.read_csv('test.csv')
```

```python
from catboost import CatBoostClassifier

model = CatBoostClassifier(iterations = 500, learning_rate = 0.15, max_depth = 6,rsm=0.87, random_state = 42, cat_features = ['AgeGroup','DoctorInCharge','EyeColor','Gender','Hypertension','EducationLevel','Diabetes','Depression','Ethnicity','FamilyHistoryParkinsons','Smoking'])
model.fit(train.drop(['PatientID','Diagnosis'],axis=1),train['Diagnosis'])
```

```python
subtask3_answer = model.predict(test.drop('PatientID',axis=1))
```

```python
subtask1_answer = [int((test['Hypertension'].iloc[x] == 1)) + int(test['Diabetes'].iloc[x] == 1) + int(test['BMI'].iloc[x] > 30.0) for x in range(0,len(test))]
subtask2_answer = [int(test['Smoking'].iloc[x] == 1) + int(test['AlcoholConsumption'].iloc[x] > 2.0) + int(test['PhysicalActivity'].iloc[x] < 1.0) for x in range(0,len(test))]
```

```python
tr1 = pd.DataFrame({
    'PatientID':test['PatientID'].tolist(),
    'subtaskID':'Task1',
    'Answer':subtask1_answer
})
tr2 = pd.DataFrame({
    'PatientID':test['PatientID'].tolist(),
    'subtaskID':'Task2',
    'Answer':subtask2_answer
})
tr3 = pd.DataFrame({
    'PatientID':test['PatientID'].tolist(),
    'subtaskID':'Task3',
    'Answer':subtask3_answer
})
pd.concat([tr1,tr2,tr3]).to_csv('submission.csv',index=False)
```
