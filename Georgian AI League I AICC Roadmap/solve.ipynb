```python
import numpy as np
import pandas as pd
import joblib

modelA = joblib.load('model_A.joblib')
modelB = joblib.load('model_B.joblib')

train = pd.read_csv('train_data.csv')
```

```python
modelApredictions = modelA.predict_proba(train.drop(['row_id','y'],axis=1).to_numpy())
modelBpredictions = modelB.predict_proba(train.drop(['row_id','y'],axis=1).to_numpy())
```

```python
answer = []
for i in range(0,len(modelApredictions)):
    lossA = -np.log(modelApredictions[i][train['y'].iloc[i]])
    lossB = -np.log(modelBpredictions[i][train['y'].iloc[i]])
    if lossA < lossB:
        answer.append(0)
    else:
        answer.append(1)
```

```python
pd.DataFrame({
    'subtaskID':1,
    'datapointID':train['row_id'].to_list(),
    'answer':answer
}).to_csv('submission.csv',index=False)
```
