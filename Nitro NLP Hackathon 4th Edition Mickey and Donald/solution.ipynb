```python
import numpy as np
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
from catboost import CatBoostClassifier
train = pd.read_csv('train_data.csv')
test = pd.read_csv('test_data.csv')
```

```python
sub1_model = CatBoostClassifier(iterations = 450, learning_rate = 0.16, l2_leaf_reg = 0.25, rsm = 0.89, text_features = ['sample'], random_state = 42, max_depth = 8)
sub1_model.fit(train.drop(['datapointID','dialect','category'],axis=1), train['dialect'].tolist())
```

```python
subtask1_answer = sub1_model.predict(test.drop(['datapointID'],axis=1))
```

```python
sub2_model = CatBoostClassifier(iterations = 750, learning_rate = 0.18, l2_leaf_reg = 0.25, rsm = 0.89, text_features = ['sample'], random_state = 42, max_depth = 8)
sub2_model.fit(train.drop(['datapointID','dialect','category'],axis=1), train['category'].tolist())
```

```python
subtask2_answer = sub2_model.predict(test.drop(['datapointID'],axis=1))[:,0]
```

```python
pd.DataFrame({
    'subtaskID': [1] * len(test) + [2] * len(test),
    'datapointID': test['datapointID'].tolist() + test['datapointID'].tolist(),
    'answer':list(subtask1_answer) + list(subtask2_answer)
}).to_csv('submissio.csv',index=False)
```
